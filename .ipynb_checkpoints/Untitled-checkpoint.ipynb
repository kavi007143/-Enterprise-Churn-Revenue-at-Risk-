```python
import pandas as pd 
import os

path = 'Extracted_churn_analytic'

df_accounts = pd.read_csv(os.path.join(path,'ravenstack_accounts.csv'))
df_churn = pd.read_csv(os.path.join(path,'ravenstack_churn_events.csv'))
df_usage = pd.read_csv(os.path.join(path,'ravenstack_feature_usage.csv'))
df_subs = pd.read_csv(os.path.join(path,'ravenstack_subscriptions.csv'))
df_tickets = pd.read_csv(os.path.join(path,'ravenstack_support_tickets.csv'))
print('All files are loaded successfully!')
```

```python
print(df_accounts.shape)
print(df_churn.shape)
print(df_usage.shape)
print(df_subs.shape)
print(df_tickets.shape)
```

```python
#Aggregation & Feature Engineering
```

```python
import pandas as pd

#Aggregate support tickets
#calculate,How many ticket to collect the companies.calculate the time

df_tickets_agg = df_tickets.groupby('account_id').agg(
    total_tickets=('ticket_id','count'),
    avg_resolution_time=('resolution_time_hours', 'mean'),
    avg_satisfaction_score=('satisfaction_score','mean'),
    total_escalations=('escalation_flag','sum')
).reset_index()

#Aggregate Subscriptions
#how much money to spent the companies to seprate companies,how many seats to collect

df_subs_agg = df_subs.groupby('account_id').agg(
    total_subscriptions=('subscription_id','count'),
    total_mrr=('mrr_amount','sum'),
    total_arr=('arr_amount','sum'),
    total_seats_purchased=('seats','sum'),
    total_upgrades=('upgrade_flag','sum'),
    total_downgrades=('downgrade_flag','sum')
).reset_index()

#merge everything for accounts

df_master = pd.merge(
    df_accounts,
    df_subs_agg,
    on='account_id',
    how='left'
)

df_master = pd.merge(
    df_master,
    df_tickets_agg,
    on='account_id',
    how='left'
)

#no subscriptions
df_master['total_subscriptions'] = df_master['total_subscriptions'].fillna(0)
df_master['total_mrr'] = df_master['total_mrr'].fillna(0)
df_master['total_arr'] = df_master['total_arr'].fillna(0)
df_master['total_seats_purchased'] = df_master['total_seats_purchased'].fillna(0)
df_master['total_upgrades'] = df_master['total_upgrades'].fillna(0)
df_master['total_downgrades'] = df_master['total_downgrades'].fillna(0)


#cannot get ticket called Nan value is 0
df_master['total_tickets'] = df_master['total_tickets'].fillna(0)
df_master['total_escalations'] = df_master['total_escalations'].fillna(0)

#tickets related averages
df_master['avg_resolution_time'] = (
    df_master['avg_resolution_time']
        .fillna(df_master['avg_resolution_time'].mean())
)

df_master['avg_satisfaction_score'] = (
    df_master['avg_satisfaction_score']
        .fillna(5)
)


print('Ready for Machine Learning!')
print(f'\nMaster table shape(Rows,Columns):'
      f'{df_master.shape}')

print('\nColumns:')
print(list(df_master.columns))

print('\nMissing values:')
print(df_master.isnull().sum())
```

```python
#data preprocessing & data spliting
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import pandas as pd

x = df_master.drop(columns=['account_id','account_name','signup_date','churn_flag'])
y = df_master['churn_flag']

x_encoded = pd.get_dummies(x, columns = ['industry','country','referral_source','plan_tier'], drop_first = True)
x_encoded = x_encoded.astype(float)

x_train,x_test,y_train,y_test = train_test_split(x_encoded, y, test_size=0.2, random_state=42, stratify=y)

print('data preprocessing and data spliting completed!')
print(f'x_train shape(Training Features): {x_train.shape}')
print(f'x_test shape(Testing Features): {x_test.shape}')
print(f'\nTarget Variable churn Distribution in Training Set:\n{y_train.value_counts()}')
```

```python
#create module and evaluation
```

```python
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import classification_report,roc_auc_score

print('modle training process, waiting...\n')

rf_model = RandomForestClassifier(n_estimators=100, class_weight= 'balanced',random_state=42)
rf_model.fit(x_train,y_train)
rf_preds = rf_model.predict(x_test)

xgb_model = XGBClassifier(scale_pos_weight=3.54,random_state=42,eval_metric='logloss')
xgb_model.fit(x_train,y_train)
xgb_preds = xgb_model.predict(x_test)

print('===== RANDOM FOREST REPORT =====')
print(classification_report(y_test,rf_preds))
print(f'ROC-AUC Score: {roc_auc_score(y_test, rf_model.predict_proba(x_test)[:,1]):.2f}\n')

print('===== XGBOOST REPORT =====')
print(classification_report(y_test, xgb_preds))
print(f'ROC-AUC-Score: {roc_auc_score(y_test, xgb_model.predict_proba(x_test)[:,1]):.2f}')
```

```python
#Hyperparameter Tuning & Smote-like Class weights
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, roc_auc_score

print('XGBOOST Tune Model processing(Hyperparameter Tuning)...\n')
xgb_tuned = XGBClassifier(
    scale_pos_weight=6.0,
    max_depth=3,
    learning_rate=0.05,
    n_estimators=150,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric='logloss'
)

xgb_tuned.fit(x_train,y_train)
tuned_preds = xgb_tuned.predict(x_test)

print('===== TUNED XGBOOST REPORT =====')
print(classification_report(y_test, tuned_preds))
print(f'ROC-AUC-Score: {roc_auc_score(y_test, xgb_tuned.predict_proba(x_test)[:,1]):2f}')
```

```python
#B2B Revenue-at-Risk Analysis(The Executive Touch)
```

```python
# ============================================
# B2B REVENUE-AT-RISK ANALYSIS
# ============================================

# Original customer details from df_master
df_risk_analysis = df_master.loc[y_test.index].copy()

# Actual churn
df_risk_analysis['Actual_Churn'] = y_test.values

# Predicted churn
df_risk_analysis['Predicted_Churn'] = tuned_preds

# Churn probability
df_risk_analysis['Churn_Probability'] = (
    xgb_tuned.predict_proba(x_test)[:, 1]
)

print("Risk Analysis Data Created Successfully!")

print("\nRisk Analysis Shape:")
print(df_risk_analysis.shape)

print("\nRisk Analysis Columns:")
print(df_risk_analysis.columns.tolist())

# ============================================
# HIGH RISK CUSTOMERS
# ============================================

df_high_risk = df_risk_analysis[
    df_risk_analysis['Predicted_Churn'] == 1
]

print("Number of High-Risk Customers:")
print(len(df_high_risk))

# ============================================
# FINANCIAL REVENUE AT RISK
# ============================================

total_revenue_at_risk = df_high_risk['total_arr'].sum()

if len(df_high_risk) > 0:
    average_churn_probability = (
        df_high_risk['Churn_Probability'].mean()
    )
else:
    average_churn_probability = 0


print("==============================================")
print("       FINANCIAL REVENUE AT RISK REPORT")
print("==============================================")

print(
    f"High-Risk Customers       : {len(df_high_risk)}"
)

print(
    f"Revenue at Risk           : "
    f"${total_revenue_at_risk:,.2f}"
)

print(
    f"Average Churn Probability : "
    f"{average_churn_probability * 100:.2f}%"
)

print("==============================================")


# ============================================
# TOP 5 HIGH-VALUE ACCOUNTS AT RISK
# ============================================

print("\nTop 5 High-Value Accounts at Risk:")

columns_to_show = [
    'account_id',
    'total_arr',
    'total_tickets',
    'avg_satisfaction_score',
    'Churn_Probability'
]

top_5_risk_accounts = (
    df_high_risk[
        columns_to_show
    ]
    .sort_values(
        by='total_arr',
        ascending=False
    )
    .head(5)
)

print(top_5_risk_accounts)
```

```python
import  matplotlib.pyplot as plt
import numpy as np

importances = xgb_tuned.feature_importances_
features = x_train.columns

indices = np.argsort(importances)[::-1][:10]

plt.figure(figsize=(10,6))
plt.title('Top 10 Drivers of Customer Churn (B2B SaaS)', fontsize=14, fontweight='bold')
plt.barh(range(10),importances[indices],color='crimson',align='center')
plt.yticks(range(10),[features[i] for i in indices])
plt.xlabel('Relative Importance Score', fontsize=12)
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
